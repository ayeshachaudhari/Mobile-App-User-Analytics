 Final Combined Analysis


```python
 # Users + Sessions

user_sessions = users.merge(
    sessions,
    on='user_id',
    how='left'
)

user_sessions.head()
```

```python
# Users + Purchases

user_purchases = users.merge(
    purchases,
    on='user_id',
    how='left'
)

user_purchases.head()
```

```python
# Country-wise Revenue

country_revenue = user_purchases.groupby('country')['amount'].sum()

print(country_revenue.sort_values(ascending=False))
```

```python
# Revenue by Country

country_revenue.sort_values(ascending=False).plot(kind='bar')

plt.title('Revenue by Country')
plt.xlabel('Country')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.show()
```

```python
# Average Session Duration by Country

avg_session_country = user_sessions.groupby('country')['duration_min'].mean()

print(avg_session_country.sort_values(ascending=False))
```

```python
avg_session_country.sort_values(ascending=False).plot(kind='bar')

plt.title('Average Session Duration by Country')
plt.xlabel('Country')
plt.ylabel('Average Duration (Minutes)')
plt.xticks(rotation=45)
plt.show()
```

️Users + Sessions + Events + Purchases combine 

```python
final_data = (
    users
    .merge(sessions, on='user_id', how='left')
    .merge(events, on='user_id', how='left')
    .merge(purchases, on='user_id', how='left')
)
```

```python
final_data.shape
```

```python
final_data.shape
```

```python
# Table 1 — Country Summary

country_summary = users.groupby('country').agg(
    total_users=('user_id', 'nunique')
).reset_index()

country_summary
```

```python
# Table 2 — Session Summary

session_summary = sessions.groupby('device').agg(
    total_sessions=('session_id', 'nunique'),
    average_duration=('duration_min', 'mean')
).reset_index()

session_summary['average_duration'] = session_summary['average_duration'].round(2)

session_summary
```

```python
# Table 3 — Event Summary
event_summary = events.groupby('event_type').agg(
    total_events=('event_id', 'nunique')
).reset_index()

event_summary
```

```python
# Table 4 — Revenue Summary

revenue_summary = purchases.groupby('product').agg(
    total_revenue=('amount', 'sum'),
    average_purchase=('amount', 'mean')
).reset_index()

revenue_summary['average_purchase'] = revenue_summary['average_purchase'].round(2)

revenue_summary
```

```python
# Table 5 — Country Revenue

country_revenue = (
    users[['user_id', 'country']]
    .merge(purchases, on='user_id', how='inner')
    .groupby('country')
    .agg(
        total_revenue=('amount', 'sum'),
        total_purchases=('purchase_id', 'nunique')
    )
    .reset_index()
)

country_revenue
```

```python
country_summary.to_csv('country_summary.csv', index=False)

session_summary.to_csv('session_summary.csv', index=False)

event_summary.to_csv('event_summary.csv', index=False)

revenue_summary.to_csv('revenue_summary.csv', index=False)

country_revenue.to_csv('country_revenue.csv', index=False)
```
